# 2. Hafta Ödevi: Öğrenci Sınav Notu Analiz Sistemi

**Menti:** Burçin Nur Alkan  
**Konu:** If/Else, Döngüler, Fonksiyonlar, Try-Except, Koşullu Filtreleme

Bu ödevde bir sınıfın vize notlarını analiz eden basit bir sistem yazdım:
- Notları harf notuna çeviriyorum (AA, BA, BB, ...)
- Belirli bir notun üzerindekileri filtreliyorum
- Geçen / kalan öğrencileri ayırıyorum
- Sınıf istatistiklerini (ortalama, en yüksek, en düşük) çıkarıyorum
- Kullanıcı hatalarını `try-except` ile yakalıyorum

## 1. Veri: Öğrenci Notları

Öğrenci adı → sınav notu şeklinde bir sözlük kullanıyorum. Geçme notunu en başta sabit olarak tanımladım.

```python
# Gecme notu (tek yerden yonetiliyor)
GECME_NOTU = 50

notlar = {
    "Ayse": 92,
    "Mehmet": 47,
    "Zeynep": 78,
    "Can": 65,
    "Elif": 33,
    "Burak": 88,
    "Deniz": 55,
    "Selin": 71,
}

print("Sinif not listesi:")
for ogrenci, not_ in notlar.items():
    print(f"  {ogrenci}: {not_}")
```

## 2. Harf Notu Hesaplama (If / Elif / Else)

Sayısal notu klasik harf notuna çeviren bir fonksiyon. `if-elif-else` zincirinin güzel bir örneği.

```python
def harf_notu(puan):
    """Sayisal notu harf notuna cevirir."""
    if puan >= 90:
        return "AA"
    elif puan >= 80:
        return "BA"
    elif puan >= 70:
        return "BB"
    elif puan >= 60:
        return "CB"
    elif puan >= 50:
        return "CC"
    else:
        return "FF (Kaldi)"


# Her ogrenci icin harf notunu yazdiralim
for ogrenci, puan in notlar.items():
    print(f"{ogrenci:<8}: {puan} -> {harf_notu(puan)}")
```

## 3. Koşullu Filtreleme (Döngü ile)

Belirli bir eşiğin üzerinde not alan öğrencileri döndüren fonksiyon.

```python
def esik_ustu_ogrenciler(notlar, esik):
    """Notu esik degerinin uzerinde olan ogrencileri liste olarak dondurur."""
    sonuc = []
    for ogrenci, puan in notlar.items():
        if puan > esik:
            sonuc.append(ogrenci)
    return sonuc


print("70 uzeri alanlar:", esik_ustu_ogrenciler(notlar, 70))
print("85 uzeri alanlar:", esik_ustu_ogrenciler(notlar, 85))
```

## 4. Geçen / Kalan Ayrımı (`filter` + `lambda`)

Aynı işi fonksiyonel programlama ile de yapıyorum.

```python
def gecenler_kalanlar(notlar):
    """filter + lambda ile ogrencileri gecen ve kalan olarak ayirir."""
    gecenler = list(filter(lambda o: notlar[o] >= GECME_NOTU, notlar))
    kalanlar = list(filter(lambda o: notlar[o] < GECME_NOTU, notlar))
    return gecenler, kalanlar


gecen, kalan = gecenler_kalanlar(notlar)
print("Gecenler:", gecen)
print("Kalanlar:", kalan)

# map ile gecen ogrencilere "Basarili" etiketi ekleyelim
etiketli = list(map(lambda isim: f"{isim} (Basarili)", gecen))
print("Etiketli gecenler:", etiketli)
```

## 5. Sınıf İstatistikleri

Sınıfın genel durumunu özetleyen bir rapor.

```python
def sinif_raporu(notlar):
    """Sinif ortalamasi, en yuksek/dusuk not ve gecme oranini yazdirir."""
    puanlar = list(notlar.values())
    ortalama = sum(puanlar) / len(puanlar)
    en_yuksek = max(notlar, key=notlar.get)
    en_dusuk = min(notlar, key=notlar.get)
    gecen, kalan = gecenler_kalanlar(notlar)
    gecme_orani = len(gecen) / len(notlar) * 100

    print("\n=== SINIF RAPORU ===")
    print(f"Ogrenci sayisi : {len(notlar)}")
    print(f"Sinif ortalamasi: {ortalama:.1f}")
    print(f"En yuksek not   : {en_yuksek} ({notlar[en_yuksek]})")
    print(f"En dusuk not    : {en_dusuk} ({notlar[en_dusuk]})")
    print(f"Gecme orani     : %{gecme_orani:.0f}")
    print("====================\n")


sinif_raporu(notlar)
```

## 6. İnteraktif Menü — `try-except` ile Hata Yönetimi

Kullanıcı sayı yerine yanlış bir şey girerse program çökmüyor, uyarı verip tekrar soruyor.

```python
def sayi_al(mesaj):
    """Kullanicidan guvenli sekilde tam sayi alir; hatali girdide tekrar sorar."""
    while True:
        try:
            return int(input(mesaj))
        except ValueError:
            print("! Lutfen gecerli bir SAYI girin.")


def menu(notlar):
    while True:
        print("\n1) Not sorgula  2) Esik ustu filtrele  3) Sinif raporu  4) Cikis")
        secim = sayi_al("Seciminiz: ")

        if secim == 1:
            isim = input("Ogrenci adi: ")
            if isim in notlar:
                puan = notlar[isim]
                print(f"{isim}: {puan} -> {harf_notu(puan)}")
            else:
                print("! Boyle bir ogrenci yok.")
        elif secim == 2:
            esik = sayi_al("Esik notu: ")
            print(f"{esik} uzeri alanlar: {esik_ustu_ogrenciler(notlar, esik)}")
        elif secim == 3:
            sinif_raporu(notlar)
        elif secim == 4:
            print("Cikis yapildi. Iyi gunler!")
            break
        else:
            print("! 1, 2, 3 veya 4 girmelisiniz.")


# Menuyu calistirmak icin asagidaki satirin basindaki # isaretini kaldir:
# menu(notlar)
```

## 7. Özet

Bu ödevde kullandığım 2. hafta konuları:

| Konu | Nerede kullandım |
|------|------------------|
| `if / elif / else` | `harf_notu()` fonksiyonunda not aralıkları |
| `for` döngüsü | Not listesini ve filtrelemeyi dönerken |
| `while` döngüsü | Menüde ve güvenli sayı almada |
| Fonksiyonlar | `harf_notu`, `sinif_raporu`, `menu` vb. |
| `try-except` | `sayi_al()` içinde hatalı girdiyi yakalarken |
| Koşullu filtreleme | `esik_ustu_ogrenciler`, `gecenler_kalanlar` |
| `filter` / `map` / `lambda` | Geçen/kalan ayrımı ve etiketleme |

**Öğrendiğim önemli nokta:** Aynı filtreleme işini hem klasik döngüyle hem de `filter/lambda` ile yapabiliyorum; ikincisi daha kısa ama ilki daha okunaklı olabiliyor.
