## 1. Değişkenler ve Temel Veri Tipleri

Python'da değişken tanımlamak için tip belirtmeye gerek yok; Python tipi kendisi anlar.

```python
# Tam sayi (integer)
yas = 22

# Ondalikli sayi (float)
boy = 1.55

# Metin (string)
isim = "Burcinnur"

# Mantiksal deger (boolean) - sadece True veya False
ogrenci_mi = True

# type() fonksiyonu bir degiskenin tipini gosterir
print("yas degiskeninin tipi:", type(yas))
print("boy degiskeninin tipi:", type(boy))
print("isim degiskeninin tipi:", type(isim))
print("ogrenci_mi degiskeninin tipi:", type(ogrenci_mi))
```

```python
# f-string ile degiskenleri bir metin icinde kullanabiliriz
print(f"Merhaba, ben {isim}. {yas} yasindayim ve boyum {boy} metre.")
```

## 2. Listeler (list)

Liste, birden fazla veriyi sirali olarak tutan bir yapidir. Koseli parantez `[ ]` ile tanimlanir.  
**Onemli:** Listeler degistirilebilir (mutable) - eleman ekleyip cikarabiliriz.

```python
# Sevdigim programlama dilleri listesi
diller = ["Python", "Java", "SQL"]
print("Baslangic listesi:", diller)

# Listeye yeni eleman ekleme
diller.append("JavaScript")
print("Ekleme sonrasi:", diller)

# Listeden eleman cikarma
diller.remove("Java")
print("Cikarma sonrasi:", diller)

# Indeks ile elemana erisim (Python 0'dan baslar!)
print("Ilk eleman:", diller[0])
print("Son eleman:", diller[-1])

# Listenin uzunlugu
print("Listede kac dil var:", len(diller))
```

## 3. Sozlukler (dictionary)

Sozluk, verileri **anahtar-deger** (key-value) ciftleri halinde tutar. Suslu parantez `{ }` ile tanimlanir.  
Listeden farki: elemanlara sira numarasiyla degil, anahtar ismiyle ulasiriz.

```python
# Bir ogrenciyi temsil eden sozluk
ogrenci = {
    "isim": "Burcinnur",
    "bolum": "Bilgisayar Muhendisligi",
    "yas": 22,
    "dersler": ["Veri Bilimi", "Makine Ogrenmesi"]
}

# Anahtar ile degere erisim
print("Isim:", ogrenci["isim"])
print("Bolum:", ogrenci["bolum"])
print("Dersler:", ogrenci["dersler"])

# Yeni anahtar-deger ekleme
ogrenci["sehir"] = "Antalya"
print("\nSehir eklendikten sonra tum sozluk:")
print(ogrenci)
```
