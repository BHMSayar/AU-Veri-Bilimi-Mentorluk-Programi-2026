```python
import numpy as np
import time
```

```python
SIZE = 1000000
```

```python
pyList = list(range(SIZE))
npList = np.arange(SIZE)
```

Python List vs Numpy List Speed Test

Addition Phase

```python
start1 = time.time()
py_result = []
for i in range(SIZE):
    py_result.append(pyList[i] + pyList[i])
end1 = time.time()
loop_time = end1 - start1
```

```python
start2 = time.time()
numpy_result = npList + npList
end2 = time.time()
numpy_time = end2 - start2
```

```python
print("Addition Phase Result")
print(f"Loop Time : {loop_time}")
print(f"Numpy Time : {numpy_time}")
print(f"NumPy is {loop_time / numpy_time:.1f} times faster")
```

Squaring Phase

```python
start3 = time.time()
numpy_result = [x ** 2 for x in pyList]
end3 = time.time()
loop_time = end3 - start3
```

```python
start4 = time.time()
numpy_result = npList ** 2
end4 = time.time()
numpy_time = end4 - start4
```

```python
print("Squaring Phase Result")
print(f"Loop Time : {loop_time}")
print(f"Numpy Time : {numpy_time}")
print(f"NumPy is {loop_time / numpy_time:.1f} times faster")
# Difference is low bcs append with for loop is optimized
```

Sum Phase

```python
start5 = time.time()
sum = 0
for x in pyList:
    sum += x
end5 = time.time()
loop_time = end5 - start5
```

```python
start6 = time.time()
numpy_sum = np.sum(npList)
end6 = time.time()
numpy_time = end6 - start6
```

```python
print("Sum Phase Result")
print(f"Loop Time : {loop_time}")
print(f"Numpy Time : {numpy_time}")
print(f"NumPy is {loop_time / numpy_time:.1f} times faster")
```
