```python
# Gerekli Kütüphaneler
import pandas as pd
import numpy as np
```

```python
# Veriyi DataFrame olarak çekmek
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
# NaN durumu
df.isnull().sum()
```

```python
# Cabin sütununda çok fazla NaN var ondan sutünu siliyorum
df = df.drop('Cabin', axis=1)
```

```python
df.isnull().sum()
```

```python
# Age içinde çok fazla NaN o yüzden medyan ile doldurup veriyi daha kararlı hale getiriyorum
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df.isnull().sum()
```

```python
# Embarked'da sadece 2 satır eksik bu yüzden direkt boş satırları siliyorum
df = df.dropna(subset=['Embarked'])
```

```python
df.isnull().sum()
```
