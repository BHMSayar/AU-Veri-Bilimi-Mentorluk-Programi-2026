```python
def harf_notu_hesapla(ortalama):
    if ortalama < 0 or ortalama > 100:
        return "Hata: Ortalama 0 ile 100 arasında olmalıdır!"
    elif ortalama >= 90:
        return "AA - Geçti (Mükemmel)"
    elif ortalama >= 85:
        return "BA - Geçti (Çok İyi)"
    elif ortalama >= 80:
        return "BB - Geçti (İyi)"
    elif ortalama >= 75:
        return "CB - Geçti (Orta-İyi)"
    elif ortalama >= 70:
        return "CC - Geçti (Orta)"
    elif ortalama >= 65:
        return "DC - Koşullu Geçti"
    elif ortalama >= 60:
        return "DD - Koşullu Geçti"
    elif ortalama >= 50:
        return "FD - Kaldı"
    else:
        return "FF - Kaldı"
```

```python
def not_hesaplama_programi():
    while True:
        try:
            vize_girisi = input("\nVize notunuzu girin (Çıkış için 'q'): ")

            if vize_girisi.lower() == 'q':
                print("Programdan çıkılıyor.")
                break 
                
            vize = float(vize_girisi)

            final_girisi = input("Final notunuzu girin: ")
            final = float(final_girisi)

            if (vize < 0 or vize > 100) or (final < 0 or final > 100):
                print("Uyarı: Vize ve Final notları 0 ile 100 arasında olmalıdır. Lütfen tekrar deneyin.")
                continue 

            ortalama = (vize * 0.40) + (final * 0.60)

            sonuc = harf_notu_hesapla(ortalama)

            print(f"\n=> Dönem Ortalamanız: {ortalama:.2f}")
            print(f"=> Harf Notunuz ve Durumunuz: {sonuc}")

        except ValueError:
            print("Hata: Lütfen geçerli bir sayı giriniz!")
```

```python
not_hesaplama_programi()
```
